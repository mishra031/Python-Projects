```python
#tuple packing
studentData="Aman","Mca AI",78.65
print(type(studentData))

#tuple Unpacking
name,course,avgMarks=studentData
print(name)
```

```python
record="Amit","MCA",23,True
print(record[0])
print(record[1])
print(record)

name,course,age,att=record    #tuple unpacking
print(name)

#unpacking using *
# * in tuple is called extend unpacking. it allows one variable to collect multiple remaining values.
n, *ab= record
# * the operator collects the remaining elements into a list.
print(n)
print(ab)
```

```python
record="Amit", "MCA",23, True
name, *ab,att=record
print(name)
print(ab)
print(att)
```
