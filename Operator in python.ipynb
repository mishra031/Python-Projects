```python
#relational opererator/ Comparison operator in Python. Relational operator are used to compare two values.
#THe result of relationship operator is always a Boolean Value: True or False
x=10
y=20
print(x==y)
print(x<=y)
print(x>=y)
print(x>y)
print(x<y)
print(x!=y)
```

```python
#Logical Operator
x=10
y=20
print(x<y and y>x)
print(x>y and y<x)
print(x>y and y==x)
print(x<y and y==x)
```

```python
#Membership Operator
#Used to check whether a particuler value exixts inside a collection or senquence.
data=[10,20,30,40]
print(10  in data)
print(20 not in data)
```
