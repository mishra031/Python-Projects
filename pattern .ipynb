```python
#Pattern 1: Right triangle
n=5
for i in range(1,n+1):
    print("* "*i)
```

```python
#Pattern 2: Inverted triangle
n=5
for i in range(n,0,-1):
    print("* " * i)
```

```python
#Pattern 3: Pyramid
n=5
for i in range(0,n,-1):
    print(i* "  *")    
```
